```python
loaded_data = np.load(f'lead_grid.npz')
loaded_data = np.load('emittance_growth_rates.npz')

transverse_growth_rates = loaded_data['transverse_growth_rates']
longitudinal_growth_rates = loaded_data['longitudinal_growth_rates']
dp_p_values = loaded_data['dp_p_values']
emittance_list = loaded_data['emittance_list']
bunch_length_list=loaded_data['bunch_length_list']

```

Transverse cooling for all ions

```python
# Plot the heatmap for transverse growth rate
plt.figure(figsize=(10, 8))
plt.imshow(transverse_growth_rates, extent=[emittance_list.min(), emittance_list.max(), bunch_length_list.min(), bunch_length_list.max()],
           origin='lower', aspect='auto', cmap='viridis')
plt.colorbar(label=r'Transverse Emittance Growth Rate ($s^{-1}$)')
plt.xlabel(r'Normalized Emittance ($m \cdot rad$)')
plt.ylabel(r'Momentum Spread ($\Delta p/p$)')
plt.title(r'Transverse Emittance Growth Rate Heatmap')
plt.show()

# Plot the heatmap for longitudinal growth rate
plt.figure(figsize=(10, 8))
plt.imshow(longitudinal_growth_rates, extent=[emittance_list.min(), emittance_list.max(), bunch_length_list.min(), bunch_length_list.max()],
           origin='lower', aspect='auto', cmap='viridis')
plt.colorbar(label=r'Longitudinal Emittance Growth Rate ($s^{-1}$)')
plt.xlabel(r'Normalized Emittance ($m \cdot rad$)')
plt.ylabel(r'Momentum Spread ($\Delta p/p$)')
plt.title(r'Longitudinal Emittance Growth Rate Heatmap')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.colors as mcolors

# Custom binary colormap: red for positive, green for negative
cmap = mcolors.ListedColormap(['green', 'red'])
bounds = [-np.inf, 0, np.inf]
norm = mcolors.BoundaryNorm(bounds, cmap.N)

# # Plot the heatmap for transverse growth rate
# plt.figure(figsize=(10, 8))
# plt.imshow(transverse_growth_rates, extent=[emittance_list.min(), emittance_list.max(), bunch_length_list.min(), bunch_length_list.max()],
#            origin='lower', aspect='auto', cmap=cmap, norm=norm)
# plt.xlabel(r'Normalized Emittance ($m \cdot rad$)')
# plt.ylabel(r'Momentum Spread ($\Delta p/p$)')
# plt.title(r'Transverse Emittance Growth Rate Heatmap')
# plt.show()

# Plot the heatmap for longitudinal growth rate
plt.figure(figsize=(10, 8))
plt.imshow(longitudinal_growth_rates, extent=[emittance_list.min(), emittance_list.max(), bunch_length_list.min(), bunch_length_list.max()],
           origin='lower', aspect='auto', cmap=cmap, norm=norm)
plt.xlabel(r'Normalized Emittance ($m \cdot rad$)')
plt.ylabel(r'Momentum Spread ($\Delta p/p$)')
plt.title(r'Longitudinal Emittance Growth Rate Heatmap')
plt.show()
```

```python
# Example of adjusted bounds for the binary colormap
min_value = np.min(longitudinal_growth_rates)
max_value = np.max(longitudinal_growth_rates)

bounds = [min_value, 0, max_value]
norm = mcolors.BoundaryNorm(bounds, cmap.N)

# Plotting with the adjusted colormap
plt.figure(figsize=(10, 8))
plt.imshow(longitudinal_growth_rates, extent=[emittance_list.min(), emittance_list.max(), bunch_length_list.min(), bunch_length_list.max()],
           origin='lower', aspect='auto', cmap=cmap, norm=norm)
plt.xlabel(r'Normalized Emittance ($m \cdot rad$)')
plt.ylabel(r'Momentum Spread ($\Delta p/p$)')
plt.title(r'Longitudinal Emittance Growth Rate Heatmap')
plt.show()
```
