```python
import numpy as np

ion=calcium
ion_list=[lead,xenon,calcium]

for ion in ion_list:

    q0 = ion.q0
    mass0 = ion.mass0

    gamma = ion.gamma_cooling
    beta = np.sqrt(1 - 1 / gamma**2)
    
    hbar = 1.054571817e-34          # [J·s]
    eV_to_J = 1.602176634e-19       # [J/eV]

    # Energy in eV
    E_eV = ion.hw0                  # energy in eV
    #print(ion.hw0)

    # Convert to angular frequency in rad/s
    omega = (E_eV * eV_to_J) / hbar

    #print(ion.hw0/ion.gamma_cooling)

    # === Input Parameters ===
    omega_0 = omega  # [rad/s] laser frequency in ion rest frame
    sigma_delta = 2e-4  # relative momentum spread (Δp/p)
    c = constants.c     # speed of light [m/s]
    L = 25.0             # interaction length [m]

    # Derived quantities
    beta = np.sqrt(1 - 1 / gamma**2)  # relativistic beta

    # === Chirp Rate Calculation ===
    delta_omega_laser = (omega_0 * sigma_delta) / (2 * gamma * (1 + beta))
    t_interaction = L / (beta * c)
    chirp_rate = delta_omega_laser / t_interaction

    # === Output ===
    print(f"{ion.name}Chirp rate r = {chirp_rate:.5e} rad/s²")
```
