```python
import matplotlib.pyplot as plt

# Data points
years = [1975, 1986, 1988, 1990, 1995, 1995, 1996, 2003]
temperatures = [40, 1e-3, 43e-6, 3e-6, 170e-9, 2e-6, 1.1e-6, 500e-12]
labels = ['Wineland', 'Chu', 'Lett', 'Salomon', 'Wieman', 'Ketterle', 'Boiron', 'Ketterle']

# Nobel Prize winners: Chu (1986), Wieman (1995), Ketterle (1995)
nobel_years = [1986, 1995, 1995]
nobel_temperatures = [1e-3, 170e-9, 2e-6]
nobel_labels = ['Chu', 'Wieman', 'Ketterle']

doppler_limit = 240e-6  # Doppler cooling limit

# Create the plot
plt.figure(figsize=(10, 6))

# Plot all data points in blue
plt.scatter(years, temperatures, label='Other experiments', color='blue')

# Highlight Nobel Prize points in red
plt.scatter(nobel_years, nobel_temperatures, label='Nobel Prize experiments', color='red')

# Annotate each point
for i in range(len(years)):
    plt.text(years[i], temperatures[i] * 1.25, f' {labels[i]}', verticalalignment='bottom', horizontalalignment='center')

# Set y-axis to logarithmic scale
plt.yscale('log')

# Labeling
plt.xlabel('Year')
plt.ylabel('Minimum Temperature (K)')
plt.grid(True, which="both", ls="--", linewidth=0.5)

# Add a legend
plt.legend()

# Show plot
plt.tight_layout()
plt.savefig('laser_cooling_history.png', dpi=300)
plt.show()
```
